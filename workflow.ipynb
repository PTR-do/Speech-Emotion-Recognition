### WavLM + regressor head

<br> The model used as feature extractor is WavLM, that aim to build a representation from the waveform. <br>
The full model is provided with an MLP head that output the prediction of the emotion as the values of six different regression, after the feature extraction process. <br>To have a fixed input representation and more meaningful features for the task, the output of WavLM is pooled with an aggregation of three vectors <br>[mean - std - max, 2304] from the vectors of the segment.

````python
from pathlib import Path
from transformers import AutoFeatureExtractor, WavLMModel

MODEL_NAME = "microsoft/wavlm-base"
MODEL_PATH = Path("src/models/wavlm/pretrained")


if not (MODEL_PATH / "model.safetensors").exists():
    print(f"Download of {MODEL_NAME}")
    MODEL_PATH.mkdir(parents=True, exist_ok=True)

    feature_extractor = AutoFeatureExtractor.from_pretrained(MODEL_NAME)
    wavlm = WavLMModel.from_pretrained(MODEL_NAME)

    feature_extractor.save_pretrained(MODEL_PATH)
    wavlm.save_pretrained(MODEL_PATH)
    print(f"Model saved to: {MODEL_PATH}")

else:
    print(f"Model already available at: {MODEL_PATH}")


feature_extractor = AutoFeatureExtractor.from_pretrained(MODEL_PATH)
wavlm = WavLMModel.from_pretrained(MODEL_PATH)
print("WavLM loaded successfully.")
````

# Training

The experiment is conducted on the emotion recognition branch of the model. To provide a benchmark of the performance, the model will be trained with different strategies. <br> The strategies of training are the following:
- WavLm frozen, training only of the regression head;
- training with full fine tuning of WavLM;
- training with fine tuning of the last encoder layers of WavLM;
- training with fine tuning of WavLM with LoRA. 

<br>
For the four experiments will be maintened the same metrics and criterions.

````python
import torch
import torch.nn as nn
from src.training.early_stopping import EarlyStopping

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

early_stop = EarlyStopping(patience=7, min_delta=0.001)
criterion = nn.SmoothL1Loss(beta=0.03, reduction="none")

EMOTION_NAMES = [
    "happiness",
    "sadness",
    "anger",
    "fear",
    "disgust",
    "surprise",
]
````

### WavLM frozen

The first strategy of training is to adapt only the regression head, using the feature extracted from the pre-trained model WavLM. 
<br> The first step is extract and save all the features from the waveforms, in order to train only the MLP in the following steps and provide some architectural changes.

````python
from src.training.features_extraction import extract_dataset_features

FEATURES_ROOT = PROCESSED_ROOT / "features"
FEATURES_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)
train_path = FEATURES_ROOT / "train.pt"
val_path = FEATURES_ROOT / "val.pt"
test_path = FEATURES_ROOT / "test.pt"

print("\nExtraction train features")
train_features = extract_dataset_features(
    wavlm,
    device,
    train_loader,
    train_path,
)

print("\nExtraction validation features")
val_features = extract_dataset_features(
    wavlm,
    device,
    val_loader,
    val_path,
)

print("\nExtraction test features")
test_features = extract_dataset_features(
    wavlm,
    device,
    test_loader,
    test_path,
)

print("\nFEATURE EXTRACTION COMPLETED")
print(f"Train features: {train_features}")
print(f"Validation features: {val_features}")
print(f"Test features: {test_features}")
````

Features loading and creation of Dataloader for the traning of the MLP head

````python
from torch.utils.data import TensorDataset, DataLoader

FEATURES_ROOT = PROCESSED_ROOT / "features"
TRAIN_FEATURES_PATH = FEATURES_ROOT / "train.pt"
VAL_FEATURES_PATH = FEATURES_ROOT / "val.pt"
TEST_FEATURES_PATH = FEATURES_ROOT / "test.pt"

train_data = torch.load(TRAIN_FEATURES_PATH, map_location="cpu")
val_data = torch.load(VAL_FEATURES_PATH, map_location="cpu")
test_data = torch.load(TEST_FEATURES_PATH, map_location="cpu")

train_features = train_data["features"]
train_targets = train_data["targets"]

val_features = val_data["features"]
val_targets = val_data["targets"]

test_features = test_data["features"]
test_targets = test_data["targets"]

train_td = TensorDataset(train_features, train_targets)
val_td = TensorDataset(val_features, val_targets)
test_td = TensorDataset(test_features, test_targets)

train_loader_head = DataLoader(
    train_td,
    batch_size=16,
    shuffle=True,
)

val_loader_head = DataLoader(
    val_td,
    batch_size=16,
    shuffle=False,
)

test_loader_head = DataLoader(
    test_td,
    batch_size=16,
    shuffle=False,
)

print(f"Train batches: {len(train_loader_head)}")
print(f"Validation batches: {len(val_loader_head)}")
print(f"Test batches: {len(test_loader_head)}")
````

Training of the MLP head (17.856.518 trainable params)

````python
from src.models.mlp_head.emotion_regressor import EmotionRegressor
from src.training.head_training import train_model_head

regressor = EmotionRegressor().to(device)

trainable_parameters = sum(
    parameter.numel() for parameter in regressor.parameters() if parameter.requires_grad
)
print(f"Trainable parameters: {trainable_parameters:,}")

optimizer = torch.optim.AdamW(
    regressor.parameters(),
    lr=0.001,
    weight_decay=1e-05,
)

scheduler = torch.optim.lr_scheduler.ExponentialLR(optimizer=optimizer, gamma=0.9995)

(
    history_train_head,
    history_val_head,
    history_train_emotions_head,
    history_val_emotions_head,
) = train_model_head(
    model=regressor,
    device=device,
    epochs=20,
    loss_fn=criterion,
    optimizer=optimizer,
    scheduler=scheduler,
    train_dataloader=train_loader_head,
    val_dataloader=val_loader_head,
    model_path="src/models/mlp_head/checkpoints/head_train.pth",
    history_path="src/training/histories/head_training_history.pt",
    early_stopping=early_stop,
)
````

### Full Fine-tuning

````python
from src.training.end_to_end_training import train_model
from src.models.mlp_head.emotion_regressor import EmotionRegressor
from src.models.SER_branch.SER_branch import SERbranch

regressor = EmotionRegressor().to(device)
model = SERbranch(
    wavlm=wavlm,
    regressor=regressor,
).to(device)

for parameter in model.parameters():
    parameter.requires_grad = True

trainable_parameters = sum(
    parameter.numel() for parameter in model.parameters() if parameter.requires_grad
)

total_parameters = sum(parameter.numel() for parameter in model.parameters())

print(f"Total parameters: {total_parameters:,}")
print(f"Trainable parameters: {trainable_parameters:,}")

optimizer = torch.optim.AdamW(
    [
        {
            "params": model.wavlm.parameters(),
            "lr": 3e-5,
        },
        {
            "params": model.regressor.parameters(),
            "lr": 5e-4,
        },
    ],
    weight_decay=1e-5,
)

scheduler = torch.optim.lr_scheduler.ExponentialLR(
    optimizer=optimizer,
    gamma=0.9995,
)

(
    history_train,
    history_val,
    history_train_emotions,
    history_val_emotions,
) = train_model(
    model=model,
    device=device,
    epochs=20,
    loss_fn=criterion,
    optimizer=optimizer,
    scheduler=scheduler,
    train_dataloader=train_loader,
    val_dataloader=val_loader,
    model_path="src/models/full_model_SER_branch/checkpoints/full_fine_tuning.pth",
    history_path="src/training/histories/full_fine_tuning_training_history.pt",
    early_stopping=early_stop,
)
````

### Partial Fine-tuning

Partial fine-tuning of WavLM, the layer trained are the last 8/12 encoder layers, responsible of final feature representation. Regressor head is trained from scratch. 
<br>(56.707.232 + 17.856.518 trainable params)

````python
from src.training.end_to_end_training import train_model
from src.models.mlp_head.emotion_regressor import EmotionRegressor
from src.models.SER_branch.SER_branch import SERbranch

# Freeze all WavLM parameters
for param in wavlm.parameters():
    param.requires_grad = False

# Unfreeze the last 8 Transformer layers
for layer in wavlm.encoder.layers[-8:]:
    for param in layer.parameters():
        param.requires_grad = True

regressor = EmotionRegressor().to(device)
model = SERbranch(
    wavlm=wavlm,
    regressor=regressor,
).to(device)

trainable_parameters = sum(
    parameter.numel() for parameter in model.parameters() if parameter.requires_grad
)

trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
total_params = sum(p.numel() for p in model.parameters())
print(f"Trainable parameters: " f"{trainable_params:,} / {total_params:,}")

optimizer = torch.optim.AdamW(
    [
        {
            "params": model.wavlm.parameters(),
            "lr": 5e-4,
        },
        {
            "params": model.regressor.parameters(),
            "lr": 5e-4,
        },
    ],
    weight_decay=1e-5,
)

scheduler = torch.optim.lr_scheduler.ExponentialLR(
    optimizer=optimizer,
    gamma=0.9995,
)

(
    history_train,
    history_val,
    history_train_emotions,
    history_val_emotions,
) = train_model(
    model=model,
    device=device,
    epochs=20,
    loss_fn=criterion,
    optimizer=optimizer,
    scheduler=scheduler,
    train_dataloader=train_loader,
    val_dataloader=val_loader,
    model_path="src/models/full_model_SER_branch/checkpoints/last8_layers_fine_tuning.pth",
    history_path="src/training/histories/last8_layers_training_history.pt",
    early_stopping=early_stop,
)
````

### LoRA Fine-tuning

LoRA (Low Adaptation Rank) fine-tunig focused on the attention mechanism of the encoder layers. Regressor head is trained from scratch.
<br> (884.736 + 17.856.518 trainable params)

````python
from peft import LoraConfig, get_peft_model
from src.training.end_to_end_training import train_model
from src.models.mlp_head.emotion_regressor import EmotionRegressor
from src.models.SER_branch.SER_branch import SERbranch

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "k_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none",
)

wavlm = get_peft_model(wavlm, lora_config)
regressor = EmotionRegressor().to(device)
model = SERbranch(
    wavlm=wavlm,
    regressor=regressor,
).to(device)

trainable_parameters = sum(
    parameter.numel() for parameter in model.parameters() if parameter.requires_grad
)
total_parameters = sum(parameter.numel() for parameter in model.parameters())
print(f"Total parameters: {total_parameters:,}")
print(f"Trainable parameters: {trainable_parameters:,}")

optimizer = torch.optim.AdamW(
    [
        {
            "params": [
                parameter
                for name, parameter in model.named_parameters()
                if parameter.requires_grad and "lora_" in name
            ],
            "lr": 1e-4,
        },
        {
            "params": model.regressor.parameters(),
            "lr": 1e-3,
        },
    ],
    weight_decay=1e-5,
)
scheduler = torch.optim.lr_scheduler.ExponentialLR(
    optimizer=optimizer,
    gamma=0.9995,
)

(
    history_train,
    history_val,
    history_train_emotions,
    history_val_emotions,
) = train_model(
    model=model,
    device=device,
    epochs=20,
    loss_fn=criterion,
    optimizer=optimizer,
    scheduler=scheduler,
    train_dataloader=train_loader,
    val_dataloader=val_loader,
    model_path="src/models/full_model_SER_branch/checkpoints/lora_fine_tuning.pt",
    history_path="src/training/histories/lora_fine_tuning_training_history.pt",
    early_stopping=early_stop,
)
````

### MLP training with Attention Pooling

After the results of the four strategies, it is clear that the emotions in the dataset have different distribution, and some are worst to generalize for the model. <br> To have a clear vision of the experiments and the model another experiments is conducted, aiming to remove the main bottleneck of the model. 
<br> The [mean-std-max] pooling is replaced with a simple version of Attention Pooling, that produce a 768 vector composed by a weighted sum of the T vectors produced from the segment. <br> (12.156.423 trainable params)

````python
import torch
import pandas as pd
import numpy as np

EMOTION_NAMES = [
    "happiness",
    "sadness",
    "anger",
    "fear",
    "disgust",
    "surprise",
]

head_validation = torch.load(
    "src/training/histories/head_training_history.pt",
    map_location="cpu",
    weights_only=False,
)
full_validation = torch.load(
    "src/training/histories/full_fine_tuning_training_history.pt",
    map_location="cpu",
    weights_only=False,
)
last8_validation = torch.load(
    "src/training/histories/last8_layers_training_history.pt",
    map_location="cpu",
    weights_only=False,
)
lora_validation = torch.load(
    "src/training/histories/lora_fine_tuning_training_history.pt",
    map_location="cpu",
    weights_only=False,
)
head_ap_validation = torch.load(
    "src/training/histories/attention_pooling_wavlm_frozen_history.pt",
    map_location="cpu",
    weights_only=False,
)
head_lap_validation = torch.load(
    "src/training/histories/learnable_query_pooling_wavlm_frozen_history.pt",
    map_location="cpu",
    weights_only=False,
)

head_test = torch.load(
    "src/training/histories/head_train_test_history.pt",
    map_location="cpu",
    weights_only=False,
)
full_test = torch.load(
    "src/training/histories/full_fine_tuning_test_history.pt",
    map_location="cpu",
    weights_only=False,
)
last8_test = torch.load(
    "src/training/histories/last8_layers_test_history.pt",
    map_location="cpu",
    weights_only=False,
)
lora_test = torch.load(
    "src/training/histories/lora_test_history.pt",
    map_location="cpu",
    weights_only=False,
)
head_ap_test = torch.load(
    "src/training/histories/attention_pooling_test_history.pt",
    map_location="cpu",
    weights_only=False,
)
head_lap_test = torch.load(
    "src/training/histories/learnable_query_pooling_test_history.pt",
    map_location="cpu",
    weights_only=False,
)


validation_histories = {
    "Full Fine-Tuning": full_validation,
    "Last-8 Fine-Tuning": last8_validation,
    "LoRA": lora_validation,
    "Head-only": head_validation,
    "Head-only with AP": head_ap_validation,
    "Head-only with LAP": head_lap_validation,
}
test_histories = {
    "Full Fine-Tuning": full_test,
    "Last-8 Fine-Tuning": last8_test,
    "LoRA": lora_test,
    "Head-only": head_test,
    "Head-only with AP": head_ap_test,
    "Head-only with LAP": head_lap_test,
}

validation_results = []
for strategy_name, history in validation_histories.items():
    validation_losses = np.asarray(
        history["history_val"],
        dtype=float,
    )
    validation_losses_emotions = np.asarray(
        history["history_val_emotions"],
        dtype=float,
    )
    best_epoch = int(np.argmin(validation_losses))
    best_validation_loss = validation_losses[best_epoch]
    assert best_validation_loss == np.min(validation_losses), (
        f"{strategy_name}: selected epoch is not the minimum " f"validation-loss epoch."
    )
    best_validation_emotions = validation_losses_emotions[best_epoch]
    row = {
        "Strategy": strategy_name,
        "Epoch": best_epoch + 1,
        "Total Loss": best_validation_loss,
    }
    for i, emotion in enumerate(EMOTION_NAMES):
        row[emotion] = best_validation_emotions[i]
    validation_results.append(row)

test_results = []
for strategy_name, history in test_histories.items():
    test_loss = float(history["test_loss"])
    test_losses_emotions = np.asarray(
        history["test_loss_emotions"],
        dtype=float,
    )
    row = {
        "Strategy": strategy_name,
        "Total Loss": test_loss,
    }
    for i, emotion in enumerate(EMOTION_NAMES):
        row[emotion] = test_losses_emotions[i]
    test_results.append(row)


validation_df = pd.DataFrame(validation_results)
test_df = pd.DataFrame(test_results)

validation_df = validation_df[
    [
        "Strategy",
        "Epoch",
        "Total Loss",
        *EMOTION_NAMES,
    ]
]

test_df = test_df[
    [
        "Strategy",
        "Total Loss",
        *EMOTION_NAMES,
    ]
]

print("VALIDATION RESULTS — BEST EPOCH")
display(validation_df)

print("\nTEST RESULTS")
display(test_df)
````

# Inference

A proposal for the final model ready for inference is a composed model:
- a **preprocessing pipeline** with a denoising model to improve sound quality and a Voice Activity detection to find the speech region from the waveform and provide to the model a list of the segmented regions of the original audio, with indication of start and end of each segment. The model accept in input on audio, audio segment or waveform.
- a **Speech Emotion Recognition branch**, the emotion values are predicted as a **continuos contextual acustic representation**: the model predict only a maximum of 3 second per time from the segment with a short time approach, with window 3 sec and hop size 0.5 sec. 
<br>It enables the model to process longer audio sequence than the short segments of training an provide an accurate score for the whole duration.
- a **Speech recognition branch** with forced alignment word-level to timestamps.

### End-to-End Speech Emotion Recognition Pipeline

    │                       └── 6 emotion values

    ↓
Absolute timestamps
    ↓
show_results()
